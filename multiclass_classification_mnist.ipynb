```python
import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,confusion_matrix
from sklearn.metrics import precision_score,f1_score,recall_score
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/fashion-mnist_train.csv')
df.head()
```

```python
X_train,X_test,y_train,y_test = train_test_split(df.iloc[:,1:],df.iloc[:,0],test_size=0.2,random_state=2)
```

```python
X_train.shape
```

```python
logistic_regression = LogisticRegression(max_iter=100000)
decision_tree = DecisionTreeClassifier()
```

```python
logistic_regression.fit(X_train,y_train)
decision_tree.fit(X_train,y_train)
```

```python
y_pred1 = logistic_regression.predict(X_test)
y_pred2 = decision_tree.predict(X_test)
```

```python

print("Accuracy of Logistic Regression",accuracy_score(y_test,y_pred1))
print("Accuracy of Decision Trees",accuracy_score(y_test,y_pred2))
```

```python
print("Logistic Regression Confusion Matrix\n")
pd.DataFrame(confusion_matrix(y_test,y_pred1))
```

```python
print("Decision Tree Confusion Matrix\n")
pd.DataFrame(confusion_matrix(y_test,y_pred2))
```

```python

precision_score(y_test,y_pred1,average='weighted')
```

```python
recall_score(y_test,y_pred1,average='weighted')
```

```python
f1_score(y_test,y_pred1,average='weighted')
```

```python
print(classification_report(y_test,y_pred1))
```
